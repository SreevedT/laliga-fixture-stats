```python
#Mount ADLS to ADBS

if not any(mount.mountPoint == '/mnt/laliga' for mount in dbutils.fs.mounts()):
    storage_account_name = "laligadatadl"
    container_name = "rawleague"
    storage_account_access_key =  "Add your key"

    configuration = { f"fs.azure.account.key.{storage_account_name}.blob.core.windows.net": storage_account_access_key
    }

    dbutils.fs.mount(
        source = f"wasbs://{container_name}@{storage_account_name}.blob.core.windows.net/",
        mount_point = "/mnt/laliga",
        extra_configs = configuration
    )
```

```python
dbutils.widgets.text("seasonYear","")
season_year = dbutils.widgets.get("seasonYear")

df = spark.read.option("multiline", "true").json(f"/mnt/laliga/{season_year}_season.json")
df.display()


```

```python
%sql
DROP TABLE IF EXISTS laliga_season;
```

```python

%python
df_match = df.select(explode("response").alias("tournament"))\
              .select(col("tournament.fixture.id").alias("fixture_id"),
                      col("tournament.goals.home").alias("home_team_score"),
                      col("tournament.goals.away").alias("away_team_score"),
                      col("tournament.league.id").alias("league_id"),
                      col("tournament.league.name").alias("league_name"),
                      col("tournament.score.halftime.home").alias("home_team_score_ht"),
                      col("tournament.score.halftime.away").alias("away_team_score_ht"),
                      col("tournament.teams.away.id").alias("away_team_id"),
                      col("tournament.teams.away.name").alias("away_team_name"),
                      col("tournament.teams.home.id").alias("home_team_id"),
                      col("tournament.teams.home.name").alias("home_team_name"),
                      lit(f'{season_year}').alias("season_id"))
df_match.write.mode("overwrite").saveAsTable("laliga_season")

df_match.display()
```
